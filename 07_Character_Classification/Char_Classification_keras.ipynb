```python
import pandas as pd
import numpy as np
```

```python
ds = pd.read_csv("letter-recognition.data", header=None)
ds = ds.values

labels = ds[:, 0]
features = ds[:, 1:].astype(np.float32)
labels = np.array([ord(c) - 65 for c in labels])
```

```python
max_ = features.max(axis=0)
min_ = features.min(axis=0)
features = (features - min_) / (max_ - min_)
```

```python
from keras.utils import to_categorical

labels = to_categorical(labels, num_classes=26)
print(labels.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(features, labels, test_size=0.3, random_state=32)
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
from keras import Sequential
from keras import layers

model = Sequential([
    layers.Input(shape=(16, )),
    layers.Dense(32, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(32, activation="relu"),
    layers.Dense(26, activation="softmax"),
])

model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["acc"])
model.fit(x_train, y_train, batch_size=16, epochs=10, validation_data=(x_test, y_test))
```

```python
from sklearn.metrics import classification_report

y_pred = np.argmax(model.predict(x_test), axis=1)
y_test_int = np.argmax(y_test, axis=1)

print(classification_report(y_pred, y_test_int))
```
